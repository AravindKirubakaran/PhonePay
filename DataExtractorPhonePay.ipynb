```python
import pandas as pd
from sqlalchemy import create_engine
from getpass import getpass # Recommended for securely entering passwords

# # --- 1. Create Example DataFrame (Same as before) ---
# data = {'col1': [1, 2, 3], 
#         'col2': ['A', 'B', 'C'], 
#         'date': pd.to_datetime(['2024-01-01', '2024-01-02', '2024-01-03'])}
# df = pd.DataFrame(data)

# --- 2. Define MySQL Connection Parameters ---
MYSQL_USER = "root"          # e.g., 'root'
MYSQL_PASS = getpass("Enter MySQL Password: ") # Securely prompts for password
MYSQL_HOST = "localhost"                # Or your server's IP
MYSQL_PORT = "3306"                     # Default MySQL port
MYSQL_DB = "phonepay"         # The schema/database you want to use
```

```python
def CreateAndInsertData(TableName, DataFrame):
        
    # **CHANGE 1: Update the DATABASE_URL format for MySQL Connector**
    # Format: mysql+mysqlconnector://user:password@host:port/dbname
    DATABASE_URL = f"mysql+mysqlconnector://{MYSQL_USER}:{MYSQL_PASS}@{MYSQL_HOST}:{MYSQL_PORT}/{MYSQL_DB}"
    
    # Create the SQLAlchemy Engine
    try:
        # **CHANGE 2: Create the engine using the new MySQL URL**
        engine = create_engine(DATABASE_URL)
    except Exception as e:
        print(f"❌ Could not create the database engine: {e}")
        exit()

    
    try:
        print("Attempting to connect and store DataFrame...")
        
        DataFrame.to_sql(
            name=TableName,
            con=engine,
            if_exists='replace',  # Creates or replaces the table
            index=True,          # To write the DataFrame's index
            chunksize=1000        # Write in batches of 1000 rows
        )
        print(f"✅ DataFrame successfully stored to table '{TableName}' in MySQL.")
    
        # --- 4. Verify the Data (Optional) ---
        df_check = pd.read_sql(f"SELECT * FROM {TableName}", con=engine)
        print("\nData retrieved from database for verification:")
        print(df_check)
        
    except Exception as e:
        print(f"❌ An error occurred during database operation: {e}")
```
